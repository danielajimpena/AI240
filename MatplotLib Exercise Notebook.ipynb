## Exercise 3 - Draw Barchart




```python
car_brands = ['Toyota', 'Honda', 'Ford', 'BMW', 'Audi']
sales = [120, 85, 95, 60, 45]

plt.figure(figsize=(7, 4))
plt.bar(car_brands, sales, color='skyblue', edgecolor='black')
plt.title('Car Sales by Brand')
plt.xlabel('Car Brand')
plt.ylabel('Units Sold')
plt.show()
```

## Exercise 4 -  Draw Histogram


```python
import numpy as np

car_speeds = [45, 50, 55, 60, 62, 65, 65, 70, 72, 75, 78, 80, 85, 90, 95]

plt.figure(figsize=(7, 4))
plt.hist(car_speeds, bins=6, color='salmon', edgecolor='black')
plt.title('Distribution of Car Speeds')
plt.xlabel('Speed (mph)')
plt.ylabel('Frequency')
plt.show()
```

## Exercise 5 -  Draw Piechart



```python
car_types = ['Sedan', 'SUV', 'Truck', 'Coupe', 'Hatchback']
shares = [35, 30, 20, 10, 5]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

plt.figure(figsize=(6, 6))
plt.pie(shares, labels=car_types, autopct='%1.1f%%', startangle=140, colors=colors)
plt.title('Vehicle Type Market Share')
plt.show()
```
